```python
import random

# Step 1: Create a list of words
word_list = [
    "The", "A", "happy", "small", "beautiful",
    "cat", "dog", "student", "teacher",
    "runs", "eats", "plays", "likes",
    "in", "on", "the", "garden", "school",
    "quickly", "today"
]

# Step 2: Create a function to generate a random sentence
def generate_sentence(length):
    words = random.choices(word_list, k=length)

    # Step 3: Combine the words into a sentence
    sentence = " ".join(words)

    # Step 4: Capitalize the first letter and add a period
    sentence = sentence[0].upper() + sentence[1:] + "."

    return sentence

# Step 5: Ask the user for the sentence length
length = int(input("Enter the number of words: "))

# Step 6: Generate and display the sentence
if length > 0:
    print(generate_sentence(length))
else:
    print("Please enter a positive number.")
```

```python
#Create the get_words_from_file function with error handling
def get_words_from_file(file_path):
    try:
        # Open the file in read mode
        with open(file_path, "r") as file:
            # Read the content of the file
            content = file.read()
            # Split the content into a list of words
            words = content.split()
        return words
    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found. Please upload it or ensure the path is correct.")
        return []

# Example: call the function
words = get_words_from_file("words.txt")

if words:
    print(f"Successfully loaded {len(words)} words.")
    print(words[:10])
```

```python
def get_words_from_file(file_path):
    # Open the file in read mode
    with open(file_path, "r", encoding="utf-8") as file:
        # Read the content and split into a list of words
        words = file.read().split()
    return words
```

```python
## Exercise 2: Working with JSON

import json

json_string = '{"Name": "Apollos", "age": 30, "city": "Tel Aviv"}'

# Convert the JSON string into a Python dictionary
data = json.loads(json_string)

print(data)
print(type(data))
```

```python
# Print the values that actually exist in the 'data' dictionary
print("Name:", data["Name"])
print("Age:", data["age"])
print("City:", data["city"])
```

```python
# Initialize
data["company"] = {}
data["company"]["employee"] = {}

# Add the birth_date key
data["company"]["employee"]["birth_date"] = "1996-05-05"

# Print the updated employee dictionary
print(data["company"]["employee"])
```

```python
# Save the modified dictionary to a JSON file
with open("employee.json", "w") as file:
    json.dump(data, file, indent=4)

print("JSON data saved successfully to employee.json!")
```
